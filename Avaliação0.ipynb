Carregue uma imagem colorida própria e exiba-a com correção de canais (BGR
→RGB)

```python
import cv2
from google.colab.patches import cv2_imshow # Import necessário para o Colab

# Certifique-se de que o caminho da imagem está correto no seu ambiente do Colab
# Se você fez upload manual, o caminho geralmente é "/content/nome_da_imagem.png"
path = "/usr/img/1.png"
img = cv2.imread(path)

if img is None:
    print("Erro: Imagem não encontrada. Verifique o caminho.")
else:
    print("Original:")
    cv2_imshow(img)

    # Conversão para Tons de Cinza
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    print("Gray:")
    cv2_imshow(gray)

    # Conversão para HSV
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    print("HSV:")
    cv2_imshow(hsv)

    # Conversão para L*a*b*
    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    print("L*a*b*:")
    cv2_imshow(lab)

# Nota: cv2.waitKey(0) não é necessário no ambiente de notebook
```

 Imprima na tela os atributos: shape, size e dtype da imagem.

```python
import cv2
from google.colab.patches import cv2_imshow # Necessário para exibir imagens no Colab

# 1. Carregar a imagem
# Certifique-se de que o arquivo 'entrada.jpg' foi enviado para a aba de arquivos do Colab
img = cv2.imread("/usr/img/2.png")

# 2. Imprimir os atributos solicitados
print(f"Atributo 'shape' (Altura, Largura, Canais): {img.shape}")
print(f"Atributo 'size' (Total de pixels/elementos): {img.size}")
print(f"Atributo 'dtype' (Tipo de dado dos pixels): {img.dtype}")

# Exemplo detalhado baseado no livro
print("\n--- Detalhes Individuais ---")
print(f"Largura em pixels: {img.shape[1]}")
print(f"Altura em pixels: {img.shape[0]}")
print(f"Quantidade de canais: {img.shape[2]}")
```

 Salve umacópia da sua imagem original em um formato de compressão diferente
(ex: PNG para JPG)

```python
import cv2

# 1. Carregue a imagem original (ex: formato PNG)
# Certifique-se de que o arquivo 'imagem_original.png' esteja na pasta do Colab
img = cv2.imread("/usr/img/3.png")

# 2. Salve a imagem com uma nova extensão (ex: .jpg)
# O OpenCV aplicará a compressão correspondente ao formato JPG automaticamente
cv2.imwrite("/usr/img/imagem_convertida.jpg", img)

print("Imagem salva com sucesso em um novo formato!")
```

 Modifique uma região de 100 × 100 no canto superior para uma cor sólida e
exiba o resultado

```python
import cv2
from google.colab.patches import cv2_imshow

# 1. Carregar a imagem
img = cv2.imread("/usr/img/4.png")

# 2. Modificar a região de 100x100 no canto superior esquerdo
# No OpenCV, a ordem das cores é BGR (Azul, Verde, Vermelho)
# Vamos definir para Azul sólido: (255, 0, 0)
img[0:100, 0:100] = (255, 0, 0)

# 3. Exibir o resultado
cv2_imshow(img)
```

Realize um recorte (ROI) focando em um detalhe da sua foto e exiba apenas
o recorte

```python
import cv2
from google.colab.patches import cv2_imshow # Necessário para o Colab

# 1. Carregar a imagem original
img = cv2.imread("/usr/img/3.png")

# 2. Realizar o recorte (ROI)
# Exemplo: Recortando da linha 100 até 300 e da coluna 150 até 350
# O formato é [linhaInicial:linhaFinal, colunaInicial:colunaFinal]
recorte = img[100:200, 100:200]

# 3. Exibir apenas o recorte
print("Exibindo apenas o recorte (ROI):")
cv2_imshow(recorte)

# 4. Opcional: Salvar o recorte no disco
cv2.imwrite("/usr/img/detalhe_recortado.jpg", recorte)
```
